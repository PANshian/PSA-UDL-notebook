```python
# Routine to draw graph structure, highlighting original node (brown in fig 13.10)
# and neighborhood nodes (orange in figure 13.10)
def draw_graph_structure(adjacency_matrix, original_node, neighborhood_nodes=None):

  G = nx.Graph()
  n_node = adjacency_matrix.shape[0]
  for i in range(n_node):
    for j in range(i):
      if adjacency_matrix[i,j]:
          G.add_edge(i,j)

  color_map = []

  for node in G:
    if original_node[node]:
      color_map.append('brown')
    else:
      if neighborhood_nodes[node]:
        color_map.append('orange')
      else:
        color_map.append('white')

  nx.draw(G, nx.spring_layout(G, seed = 7), with_labels=True,node_color=color_map)
  plt.show()
```

Let's imagine that we want to form a batch for a node labelling task that consists of just node 16 in the output layer (highlighted).   The network consists of the input, hidden layer 1, hidden layer2, and the output layer.

```python
# 1. 确保邻接矩阵A和n_nodes已定义
# n_nodes = A.shape[0]
# 2. 定义输出层节点16
output_layer_nodes = np.zeros((n_nodes, 1))
output_layer_nodes[16] = 1.0
# 3. 定义采样数量
n_sample = 3
```

```python
# TODO: 采样连接到输出层16的Hidden Layer 2节点
# 1. 找到所有与节点16相连的候选节点
all_hidden2_candidates = np.where(A[16, :] > 0)[0]
# 2. 随机采样3个节点，无放回
np.random.seed(42)
sampled_hidden2 = np.random.choice(all_hidden2_candidates, size=n_sample, replace=False)
# 3. 生成Hidden Layer 2节点标记
hidden_layer2_nodes = np.zeros((n_nodes, 1))
hidden_layer2_nodes[sampled_hidden2, 0] = 1.0

# 画图
print("Hidden layer 2:")
draw_graph_structure(A, output_layer_nodes, hidden_layer2_nodes)
```

```python
# TODO: 采样连接到Hidden 2的Hidden Layer 1节点
# 1. 找到所有与采样的Hidden 2节点相连的候选节点
all_hidden1_candidates = np.where(A[sampled_hidden2, :].sum(axis=0) > 0)[0]
# 2. 排除已选节点：输出层16、采样的Hidden 2节点
exclude_indices = np.concatenate([[16], sampled_hidden2])
all_hidden1_candidates = all_hidden1_candidates[~np.isin(all_hidden1_candidates, exclude_indices)]
# 3. 随机采样3个节点，无放回
np.random.seed(42)
sampled_hidden1 = np.random.choice(all_hidden1_candidates, size=n_sample, replace=False)
# 4. 生成Hidden Layer 1节点：采样的Hidden 1 + 采样的Hidden 2（并集）
hidden_layer1_nodes = np.zeros((n_nodes, 1))
hidden_layer1_nodes[np.concatenate([sampled_hidden1, sampled_hidden2]), 0] = 1.0

# 画图
print("Hidden layer 1:")
draw_graph_structure(A, output_layer_nodes, hidden_layer1_nodes)
```
